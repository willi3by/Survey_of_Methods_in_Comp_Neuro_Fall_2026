```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import os
```

# Exploring Neuroscience Data

This notebook introduces the first steps in a computational neuroscience workflow.

Before selecting statistical tests, we need to understand:

- what type of data we have
- how the data are organized
- what patterns and variability exist

## 0. Loading Data

In computational neuroscience, data often come from multiple sources:

- behavioral experiments
- imaging experiments
- sequencing experiments
- neural recordings

Here we create a small example dataset.

```python
neural_data = pd.DataFrame({
    "animal": ["A1","A2","A3","A4","A5","A6"],
    "condition": ["Control","Control","Drug","Drug","Control","Drug"],
    "firing_rate": [12,15,22,20,14,18],
    "response_time": [1.2,1.1,0.8,0.9,1.3,1.0]
})

neural_data.head()
```

## 1. Data Types

Different types of measurements require different analysis approaches.

### 1.1 Continuous Data

Continuous data can take many possible values.

Examples:

- firing rate
- fluorescence intensity
- reaction time

```python
firing_rate = neural_data["firing_rate"]
firing_rate
```

```python
firing_rate.hist()
plt.xlabel("Firing rate")
plt.ylabel("Count")
plt.show()
```

### 1.2 Categorical Data

Categorical data describe groups or conditions.

Examples:

- genotype
- treatment group
- brain region

```python
neural_data["condition"]
```

### 1.3 Data Organization

Before analysis, it is important to understand:

- rows
- columns
- missing values
- experimental structure

```python
neural_data.info()
```

## 2. Mean

The mean summarizes the center of a dataset.

$$
\mu = \dfrac{\Sigma X}{N}
$$

```python
np.mean(firing_rate)
```

## 3. Variability

Variation is important because biological data are rarely identical across animals or trials.

```python
np.var(firing_rate), np.std(firing_rate)
```

## 4. Visualization

Plots help identify:

- trends
- variability
- possible outliers
- differences between groups

```python
neural_data.boxplot(column="firing_rate", by="condition")
plt.show()
```

## 5. Sampling and Bias

Neuroscience datasets often have limited sample sizes.

Important questions:

- How were animals selected?
- Are measurements independent?
- Could sampling bias affect interpretation?

## 6. Summary Tables

Before choosing statistical tests, summarize the data and inspect the differences.

```python
neural_data.groupby("condition").mean(numeric_only=True)
```
