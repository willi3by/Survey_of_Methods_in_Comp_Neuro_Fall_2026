## 0. Installing Python (install packages)

Recommended packages:

```
# Import packages
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 1. Variables and Types

Variables store information that can be used later in an analysis.

In neuroscience, variables may represent:

- number of neurons
- trial counts
- behavioral measurements
- experimental conditions

```
# Example variables
n_trials = 50
animal_id = "Mouse_01"
mean_firing_rate = 12.5

print(n_trials)
print(animal_id)
print(mean_firing_rate)
```

## 2. NumPy Arrays

NumPy arrays are commonly used for numerical neuroscience data.

Examples:

- neural firing rates
- calcium imaging traces
- image pixels

```
# A simple neural recording
spike_rates = np.array([5, 8, 12, 15, 10, 7])
spike_rates
```

```
# Basic operations
print(np.mean(spike_rates))
print(np.std(spike_rates))
```

## 3. Pandas DataFrames

DataFrames are useful for organized experimental data.

Rows often represent:

- animals
- trials
- cells

Columns often represent:

- measurements
- experimental groups
- metadata

```
data = pd.DataFrame({
    "animal": ["Mouse_1", "Mouse_2", "Mouse_3"],
    "condition": ["Control", "Drug", "Control"],
    "velocity": [12.2, 8.5, 11.1]
})

data
```

## 4. Indexing Data

Indexing allows us to select the data needed for analysis.

```
data["velocity"]
```

```
data[data["condition"] == "Control"]
```

## 5. Plotting Data

Visualization is one of the first steps in understanding neuroscience datasets.

```
plt.plot(spike_rates)
plt.xlabel("Time")
plt.ylabel("Firing rate")
plt.show()
```

## 6. Functions

Functions help make analyses reproducible by avoiding repeated code.

```
def calculate_average(signal):
    return np.mean(signal)

calculate_average(spike_rates)
```
